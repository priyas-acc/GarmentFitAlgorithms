```python
import random
garments_db = []
for i in range(100):
    garments_db.append({
        "ID": i,
        "Chest": random.randint(80, 120),
        "Waist": random.randint(70, 110),
        "Hip": random.randint(85, 115)
    })
user = {"Chest": 95, "Waist": 80, "Hip": 90}
```

```python
def penalty(user_val, garment_val):
    if garment_val < user_val:
        return float('inf')
    else:
        return garment_val - user_val
```

```python
def fit_score(user, garment):
    chest_pen = penalty(user["Chest"], garment["Chest"])
    waist_pen = penalty(user["Waist"], garment["Waist"])
    hip_pen   = penalty(user["Hip"], garment["Hip"])
    if float('inf') in [chest_pen, waist_pen, hip_pen]:
        return -1
    total_pen = 2*chest_pen + waist_pen + hip_pen
    score = max(0, 100 - total_pen)
    return score
```

```python
scores = []
for garment in garments_db:
    score = fit_score(user, garment)
    if score > 0:
        scores.append((garment["ID"], score))
top3 = sorted(scores, key=lambda x: x[1], reverse=True)[:3]


```

```python
print("Top 3 Best Fits:")
for gid, score in top3:
    print(f"Garment ID: {gid}, Fit Confidence: {score}")
```
