```python
class DataSanitizer:

    def __init__(self, measurements):

        self.measurements = measurements
        self.normalized = {}

    def normalize_units(self):

        for key, value in self.measurements.items():

            if value is None:
                continue

           if key == "Height":

                if value < 100:
                    self.normalized[key] = value * 2.54

                else:
                    self.normalized[key] = value
            else:

                if value < 80:
                    self.normalized[key] = value * 2.54

                else:
                    self.normalized[key] = value

        return self.normalized


    def validate_proportions(self):

        h = self.normalized.get("Height")
        c = self.normalized.get("Chest")
        w = self.normalized.get("Waist")
        hip = self.normalized.get("Hip")

        issues = []

        if h is None:
            issues.append("Missing Height")
            return issues

        if w:

            if w > h:
                issues.append("Waist larger than Height")

            if w < 0.2 * h:
                issues.append("Waist unusually small")

        if c:

            if c < 0.3 * h:
                issues.append("Chest too small")

            if c > 0.9 * h:
                issues.append("Chest unusually large")

        if hip:

            if hip < 0.3 * h or hip > 0.9 * h:
                issues.append("Hip outside expected range")

        return issues

    def estimate_missing(self):

        h = self.normalized.get("Height")

        if h:

            if self.normalized.get("ArmLength") is None:
                self.normalized["ArmLength"] = round(0.35 * h, 2)

            if self.normalized.get("LegLength") is None:
                self.normalized["LegLength"] = round(0.50 * h, 2)

        return self.normalized
```

```python
user_data = {
    "Height": 70,
    "Chest": 85,
    "Waist": 120,
    "Hip": 200,
}

sanitizer = DataSanitizer(user_data)
normalized = sanitizer.normalize_units()
print("Normalized:", normalized)
issues = sanitizer.validate_proportions()
print("Issues:", issues)
final = sanitizer.estimate_missing()
print("Final dataset:", final)
```
